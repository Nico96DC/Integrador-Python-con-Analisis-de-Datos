Ludmila tiene un negocio especializado en la venta de bicicletas. Hasta el momento registra todas sus operaciones en un cuaderno. Sin embargo, dado el éxito de su empresa, cada día le es más difícil poder tener control sobre las ventas de sus bicicletas y las ganancias que registra. Por esta razón, Ludmila necesita desarrollar un sistema que la ayude en su negocio. Como sabe que nos capacitamos en el curso en Programación Python se puso en contacto con nosotros para que la ayudemos a construir el sistema.


**Clase Bicicleteria**

Atributos:
bicicletas: lista de objetos de bicicletas
ganancias: valor total de las bicicletas vendidas
cantidad_de_ventas: cantidad de bicicletas vendidas.
Métodos:
vender_bicicleta( Bicicleta): El método vender bicicleta deberá sumar 1 a la cantidad de bicicletas vendidas, a ganancias acumular el valor de la bicicleta vendida. Para saber el valor de la bicicleta por número de serie que desea vender y
comprar_bicicleta(Bicicleta): El método agregar bicicleta deberá agregar una nueva bicicleta en la lista bicicletas.
 
**Clase Bicicleta**

Atributos:
nro_de_serie: Cadena de caracteres alfanumérica
modelo: Modelo de la bicicleta
año: Año del modelo
precio: Valor de la bicicleta
Métodos:
set_precio(precio): Debe poder modificar el precio de una bicicleta
get_precio(): Devuelve el valor del precio de una bicicleta
get_nro_de_serie(): Devuelve el nro de serie.

**Comencemos:**

```python
# a)	Crear la clase Bicicleteria.
# b)	Declarar los atributos bicicletas, ganancias y cantidad_de_ventas.
class Bicicleteria():
    def __init__(self, bicicletas, ganancias, cantidad_de_ventas):
        self.bicicletas = bicicletas
        self.ganancias = ganancias
        self.cantidad_de_ventas = cantidad_de_ventas
```

```python
# c)	Implementar el método vender_bicicleta() de la clase Bicicleteria.
# d)	Implementar el método comprar_bicicleta().
def __vender_bicicleta__(self, Bicicleta):
    if (len(self.bicicletas) == 0):
        pass
    else:
        self.ganancias += Bicicleta.precio
        self.cantidad_de_ventas += 1
        self.bicicletas.remove(Bicicleta.nro_de_serie)
def __comprar_bicicleta__(self, Bicicleta):
    self.bicicletas.append(Bicicleta.nro_de_serie)
```

```python
# e)	Crear la clase Bicicleta.
# f)	Declarar los atributos de la clase Bicicleta: nro_de_serie, modelo, año y precio.
class Bicicleta():
    def __init__(self, nro_de_serie, modelo, año):
        self.nro_de_serie = nro_de_serie
        self.modelo = modelo
        self.año = año
```

```python
# g)	Implementar el método get_precio().
# h)	Implementar el método set_precio().
# i)	Implementar el método get_nro_de_serie().
def set_precio(self, precio):
    self.precio = precio
def get_precio(self):
    return self.precio
def get_nro_de_serie(self):
    return self.nro_de_serie
```

```python
# j)	Se deberán crear al menos cuatro bicicletas y una bicicleteria. Deben usarse todos los métodos al menos una vez.
local = Bicicleteria(['Uno', 'Dos', 'Tres', 'Cuatro'], 0, 0)

bici1 = Bicicleta("Uno", "Montaña", 2013)
bici2 = Bicicleta("Dos", "Urbana", 2016)
bici3 = Bicicleta("Tres", "Híbrida", 2008)
bici4 = Bicicleta("Cuatro", "Ruta", 2017)

bici1.set_precio(23000.00)
bici1.get_precio()
bici1.get_nro_de_serie()
bici2.set_precio(11599.99)
bici2.get_precio()
bici2.get_nro_de_serie()
bici3.set_precio(13140.00)
bici3.get_precio()
bici3.get_nro_de_serie()
bici4.set_precio(32100.00)
bici4.get_precio()
bici4.get_nro_de_serie()

local.__vender_bicicleta__(bici1)
local.__vender_bicicleta__(bici3)
local.__comprar_bicicleta__(bici4)
```
